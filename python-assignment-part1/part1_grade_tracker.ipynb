```python
# Task 1 - Clean Student Data

raw_students = [
    {"name": " ayesha SHARMA ", "roll": "101", "marks_str": "88, 72, 95, 60, 78"},
    {"name": "ROHIT verma", "roll": "102", "marks_str": "55, 68, 49, 72, 61"},
    {"name": " Priya Nair ", "roll": "103", "marks_str": "91, 85, 88, 94, 79"},
    {"name": "karan MEHTA", "roll": "104", "marks_str": "40, 55, 38, 62, 50"},
    {"name": " Sneha pillai ", "roll": "105", "marks_str": "75, 80, 70, 68, 85"},
]

cleaned_students = []

print("Starting Student Data Cleaning...\n")

for student in raw_students:
    # Clean name, roll, and marks
    name = student["name"].strip().title()
    roll = int(student["roll"])
    marks = [int(m.strip()) for m in student["marks_str"].split(",")]

    # Check if name is valid (only alphabetic characters)
    is_valid = all(word.isalpha() for word in name.split())

    # Store cleaned data
    cleaned_students.append({
        "name": name,
        "roll": roll,
        "marks": marks,
        "is_valid": is_valid
    })

    # Print status - using simple text (no special symbols)
    status = "Valid" if is_valid else "Invalid"
    print(f"{name} (Roll {roll}): {status} name")

    # Print formatted profile card
    print("=" * 40)
    print(f"Student : {name}")
    print(f"Roll No : {roll}")
    print(f"Marks   : {marks}")
    print("=" * 40)
    print()

# Final requirement: Print for roll number 103
print("=== Final Output for Roll Number 103 ===")
for stu in cleaned_students:
    if stu["roll"] == 103:
        cleaned_name = stu["name"]
        print(f"ALL CAPS   : {cleaned_name.upper()}")
        print(f"lowercase  : {cleaned_name.lower()}")
        break
```

```python
# TASK 2 - Marks Analysis Using Loops & Conditionals

student_name = "Ayesha Sharma"
subjects = ["Math", "Physics", "CS", "English", "Chemistry"]
marks = [88, 72, 95, 60, 78]

print("TASK 2: Marks Analysis for", student_name)
print("-" * 50)

# Part 1: Print each subject with marks and grade using for loop
print("Subject-wise Marks and Grades:")
for i in range(len(subjects)):
    subject = subjects[i]
    mark = marks[i]

    if mark >= 90:
        grade = "A+"
    elif mark >= 80:
        grade = "A"
    elif mark >= 70:
        grade = "B"
    elif mark >= 60:
        grade = "C"
    else:
        grade = "F"

    print(f"{subject:12} : {mark:3} marks  ->  Grade: {grade}")

# Part 2: Calculate Total, Average, Highest and Lowest
total_marks = sum(marks)
average = total_marks / len(marks)

highest_mark = max(marks)
lowest_mark = min(marks)

highest_subject = subjects[marks.index(highest_mark)]
lowest_subject = subjects[marks.index(lowest_mark)]

print("\nSUMMARY")
print("=" * 50)
print(f"Total Marks      : {total_marks}")
print(f"Average Marks    : {average:.2f}")
print(f"Highest Score    : {highest_subject} ({highest_mark} marks)")
print(f"Lowest Score     : {lowest_subject} ({lowest_mark} marks)")
print("=" * 50)

# Part 3: Interactive Marks Entry System using while loop
print("\nInteractive Marks Entry System")
print("Type 'done' as subject name to stop.\n")

new_marks = []
count = 0

while True:
    subject = input("Enter subject name (or 'done' to finish): ").strip()

    if subject.lower() == "done":
        break

    if not subject:
        print("Warning: Subject name cannot be empty!\n")
        continue

    try:
        mark_input = input(f"Enter marks for {subject} (0-100): ").strip()
        mark = float(mark_input)

        if mark < 0 or mark > 100:
            print("Warning: Marks must be between 0 and 100. Skipped.\n")
            continue

        new_marks.append(mark)
        count += 1
        print(f"Added: {subject} -> {mark} marks\n")

    except ValueError:
        print("Warning: Invalid input! Please enter a valid number.\n")

# Final Updated Summary
if count > 0:
    all_marks = marks + new_marks
    updated_average = sum(all_marks) / len(all_marks)

    print("\nUPDATED SUMMARY")
    print("=" * 50)
    print(f"New subjects added : {count}")
    print(f"Original Average   : {average:.2f}")
    print(f"Updated Average    : {updated_average:.2f}")
    print("=" * 50)
else:
    print("\nNo new subjects were added.")
```

```python
# TASK 3 - Class Performance Summary

class_data = [
    ("Ayesha Sharma", [88, 72, 95, 60, 78]),
    ("Rohit Verma", [55, 68, 49, 72, 61]),
    ("Priya Nair", [91, 85, 88, 94, 79]),
    ("Karan Mehta", [40, 55, 38, 62, 50]),
    ("Sneha Pillai", [75, 80, 70, 68, 85]),
]

print("TASK 3: CLASS PERFORMANCE SUMMARY")
print("=" * 50)
print("Name                | Average | Status")
print("-" * 50)

passed = 0
failed = 0
total_avg = 0
topper_name = ""
topper_avg = -1

for name, marks in class_data:
    # Calculate average
    avg = sum(marks) / len(marks)
    avg_rounded = round(avg, 2)

    # Determine status
    status = "Pass" if avg_rounded >= 60 else "Fail"

    # Count pass/fail
    if status == "Pass":
        passed += 1
    else:
        failed += 1

    # Track class topper
    if avg_rounded > topper_avg:
        topper_avg = avg_rounded
        topper_name = name

    # Add to total for class average
    total_avg += avg_rounded

    # Print row
    print(f"{name:19} | {avg_rounded:7.2f} | {status}")

print("-" * 50)

# Final Summary
class_average = total_avg / len(class_data)

print(f"\nNumber of students who Passed : {passed}")
print(f"Number of students who Failed : {failed}")
print(f"Class Topper                  : {topper_name} ({topper_avg:.2f})")
print(f"Class Average                 : {class_average:.2f}")

```
