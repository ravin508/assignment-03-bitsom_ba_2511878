```python
menu = {
    "Paneer Tikka": {"category": "Starters", "price": 180.0, "available": True},
    "Chicken Wings": {"category": "Starters", "price": 220.0, "available": False},
    "Veg Soup": {"category": "Starters", "price": 120.0, "available": True},
    "Butter Chicken": {"category": "Mains", "price": 320.0, "available": True},
    "Dal Tadka": {"category": "Mains", "price": 180.0, "available": True},
    "Veg Biryani": {"category": "Mains", "price": 250.0, "available": True},
    "Garlic Naan": {"category": "Mains", "price": 40.0, "available": True},
    "Gulab Jamun": {"category": "Desserts", "price": 90.0, "available": True},
    "Rasgulla": {"category": "Desserts", "price": 80.0, "available": True},
    "Ice Cream": {"category": "Desserts", "price": 110.0, "available": False},
}

# ====================== TASK 1: Explore the Menu ======================

# 1. Print full menu grouped by category
print("===== RESTAURANT MENU =====\n")

# Get unique categories and sort them for consistent order (Starters -> Mains -> Desserts)
categories = sorted(set(item["category"] for item in menu.values()))

for category in categories:
    print(f"===== {category} =====")
    for item_name, details in menu.items():
        if details["category"] == category:
            status = "Available" if details["available"] else "Unavailable"
            print(f"{item_name} ₹{details['price']:.2f} [{status}]")
    print()

# 2. Computations using dictionary methods
total_items = len(menu)
available_items = sum(1 for details in menu.values() if details["available"])

# Most expensive item
most_expensive = max(menu.items(), key=lambda x: x[1]["price"])

# Items under ₹150
under_150 = {name: details["price"]
             for name, details in menu.items()
             if details["price"] < 150}

# Display results
print("=== MENU STATISTICS ===")
print(f"Total number of items on the menu     : {total_items}")
print(f"Total number of available items       : {available_items}")
print(f"Most expensive item                   : {most_expensive[0]} ₹{most_expensive[1]['price']:.2f}")
print("\nItems priced under ₹150:")
for name, price in sorted(under_150.items()):
    print(f"  • {name} ₹{price:.2f}")
```

```python
# ==================== Task 2 ====================
menu = {
    "Paneer Tikka": {"category": "Starters", "price": 180.0, "available": True},
    "Chicken Wings": {"category": "Starters", "price": 220.0, "available": False},
    "Veg Soup": {"category": "Starters", "price": 120.0, "available": True},
    "Butter Chicken": {"category": "Mains", "price": 320.0, "available": True},
    "Dal Tadka": {"category": "Mains", "price": 180.0, "available": True},
    "Veg Biryani": {"category": "Mains", "price": 250.0, "available": True},
    "Garlic Naan": {"category": "Mains", "price": 40.0, "available": True},
    "Gulab Jamun": {"category": "Desserts", "price": 90.0, "available": True},
    "Rasgulla": {"category": "Desserts", "price": 80.0, "available": True},
    "Ice Cream": {"category": "Desserts", "price": 110.0, "available": False},
}

# ========================= TASK 2: Cart Operations =========================
cart = []

def add_to_cart(item_name, quantity=1):
    if item_name not in menu:
        print(f"Error: '{item_name}' does not exist in the menu.")
        return

    if not menu[item_name]["available"]:
        print(f"Error: '{item_name}' is currently unavailable.")
        return

    # Check if item already in cart
    for item in cart:
        if item["item"] == item_name:
            item["quantity"] += quantity
            print(f"Updated: {item_name} quantity increased to {item['quantity']}")
            return

    # Add new item
    price = menu[item_name]["price"]
    cart.append({"item": item_name, "quantity": quantity, "price": price})
    print(f"Added: {item_name} x {quantity}")


def remove_from_cart(item_name):
    for i, item in enumerate(cart):
        if item["item"] == item_name:
            del cart[i]
            print(f"Removed: {item_name} from cart")
            return
    print(f"Error: '{item_name}' not found in cart.")


def print_cart():
    if not cart:
        print("Cart is empty.")
        return

    print("\n--- Current Cart ---")
    for item in cart:
        subtotal = item["quantity"] * item["price"]
        print(f"{item['item']} x{item['quantity']} = ₹{subtotal:.2f}")
    print("--------------------")


# ====================== SIMULATION ======================
print("Starting Order Simulation...\n")

add_to_cart("Paneer Tikka", 2)
print_cart()

add_to_cart("Gulab Jamun", 1)
print_cart()

add_to_cart("Paneer Tikka", 1)
print_cart()

add_to_cart("Mystery Burger")
print_cart()

add_to_cart("Chicken Wings")
print_cart()

remove_from_cart("Gulab Jamun")
print_cart()

# ====================== FINAL ORDER SUMMARY ======================
print("\n========== ORDER SUMMARY ==========")

subtotal = 0.0
for item in cart:
    item_total = item["quantity"] * item["price"]
    subtotal += item_total
    print(f"{item['item']} x{item['quantity']} ₹{item_total:.2f}")

gst = subtotal * 0.05
total_payable = subtotal + gst

print("------------------------------------")
print(f"Subtotal      : ₹{subtotal:.2f}")
print(f"GST (5%)      : ₹{gst:.2f}")
print(f"Total Payable : ₹{total_payable:.2f}")
print("====================================")
```

```python
import copy

# ==================== DATA ====================
menu = {
    "Paneer Tikka": {"category": "Starters", "price": 180.0, "available": True},
    "Chicken Wings": {"category": "Starters", "price": 220.0, "available": False},
    "Veg Soup": {"category": "Starters", "price": 120.0, "available": True},
    "Butter Chicken": {"category": "Mains", "price": 320.0, "available": True},
    "Dal Tadka": {"category": "Mains", "price": 180.0, "available": True},
    "Veg Biryani": {"category": "Mains", "price": 250.0, "available": True},
    "Garlic Naan": {"category": "Mains", "price": 40.0, "available": True},
    "Gulab Jamun": {"category": "Desserts", "price": 90.0, "available": True},
    "Rasgulla": {"category": "Desserts", "price": 80.0, "available": True},
    "Ice Cream": {"category": "Desserts", "price": 110.0, "available": False},
}

inventory = {
    "Paneer Tikka": {"stock": 10, "reorder_level": 3},
    "Chicken Wings": {"stock": 8, "reorder_level": 2},
    "Veg Soup": {"stock": 15, "reorder_level": 5},
    "Butter Chicken": {"stock": 12, "reorder_level": 4},
    "Dal Tadka": {"stock": 20, "reorder_level": 5},
    "Veg Biryani": {"stock": 6, "reorder_level": 3},
    "Garlic Naan": {"stock": 30, "reorder_level": 10},
    "Gulab Jamun": {"stock": 5, "reorder_level": 2},
    "Rasgulla": {"stock": 4, "reorder_level": 3},
    "Ice Cream": {"stock": 7, "reorder_level": 4},
}

# ========================= TASK 3: Inventory Tracker =========================

# Step 1: Create deep copy
inventory_backup = copy.deepcopy(inventory)

print("=== Deep Copy Created ===\n")

# Step 2: Demonstrate deep copy
inventory["Paneer Tikka"]["stock"] = 5
print("After changing Paneer Tikka stock to 5:")
print("inventory['Paneer Tikka']     =", inventory["Paneer Tikka"])
print("inventory_backup['Paneer Tikka'] =", inventory_backup["Paneer Tikka"])
print()

# Step 3: Restore original inventory
inventory = copy.deepcopy(inventory_backup)
print("Inventory restored to original state.\n")

# Final cart from Task 2
cart = [
    {"item": "Paneer Tikka", "quantity": 3, "price": 180.0}
]

# Step 4: Fulfill the order
print("=== Fulfilling Order ===\n")

for item in cart:
    item_name = item["item"]
    qty = item["quantity"]

    if item_name in inventory:
        current = inventory[item_name]["stock"]
        if current >= qty:
            inventory[item_name]["stock"] -= qty
            print(f"Deducted {qty} of {item_name}. Remaining: {inventory[item_name]['stock']}")
        else:
            print(f"Warning: Insufficient stock for {item_name}. Only {current} available.")
            inventory[item_name]["stock"] = 0

print("\n=== Reorder Alerts ===\n")

for item_name, data in inventory.items():
    if data["stock"] <= data["reorder_level"]:
        print(f"Reorder Alert: {item_name} — Only {data['stock']} unit(s) left (reorder level: {data['reorder_level']})")

# Final display
print("\n=== Final Inventory ===")
for item, data in inventory.items():
    print(f"{item}: {data['stock']} units (Reorder Level: {data['reorder_level']})")

print("\nBackup is unchanged → Deep copy successful!")
```

```python
# ==================== SALES LOG DATA ====================
sales_log = {
    "2025-01-01": [
        {"order_id": 1, "items": ["Paneer Tikka", "Garlic Naan"], "total": 220.0},
        {"order_id": 2, "items": ["Gulab Jamun", "Veg Soup"], "total": 210.0},
        {"order_id": 3, "items": ["Butter Chicken", "Garlic Naan"], "total": 360.0},
    ],
    "2025-01-02": [
        {"order_id": 4, "items": ["Dal Tadka", "Garlic Naan"], "total": 220.0},
        {"order_id": 5, "items": ["Veg Biryani", "Gulab Jamun"], "total": 340.0},
    ],
    "2025-01-03": [
        {"order_id": 6, "items": ["Paneer Tikka", "Rasgulla"], "total": 260.0},
        {"order_id": 7, "items": ["Butter Chicken", "Veg Biryani"], "total": 570.0},
        {"order_id": 8, "items": ["Garlic Naan", "Gulab Jamun"], "total": 130.0},
    ],
    "2025-01-04": [
        {"order_id": 9, "items": ["Dal Tadka", "Garlic Naan", "Rasgulla"], "total": 300.0},
        {"order_id": 10, "items": ["Paneer Tikka", "Gulab Jamun"], "total": 270.0},
    ],
}

# ========================= TASK 4: Daily Sales Log Analysis =========================

# 1. Print total revenue per day
print("=== Daily Revenue ===\n")
daily_revenue = {}

for date, orders in sales_log.items():
    total = sum(order["total"] for order in orders)
    daily_revenue[date] = total
    print(f"{date} : ₹{total:.2f}")

# 2. Best-selling day
best_day = max(daily_revenue, key=daily_revenue.get)
print(f"\nBest Selling Day: {best_day} (₹{daily_revenue[best_day]:.2f})")

# 3. Most ordered item
item_count = {}

for orders in sales_log.values():
    for order in orders:
        for item in order["items"]:
            if item in item_count:
                item_count[item] += 1
            else:
                item_count[item] = 1

most_ordered_item = max(item_count, key=item_count.get)
print(f"Most Ordered Item: {most_ordered_item} ({item_count[most_ordered_item]} times)")

# 4. Add new day
sales_log["2025-01-05"] = [
    {"order_id": 11, "items": ["Butter Chicken", "Gulab Jamun", "Garlic Naan"], "total": 490.0},
    {"order_id": 12, "items": ["Paneer Tikka", "Rasgulla"], "total": 260.0},
]

print("\nNew day 2025-01-05 added successfully!\n")

# 5. Reprint revenue after adding new day
print("=== Updated Daily Revenue ===\n")
daily_revenue = {}

for date, orders in sales_log.items():
    total = sum(order["total"] for order in orders)
    daily_revenue[date] = total
    print(f"{date} : ₹{total:.2f}")

best_day = max(daily_revenue, key=daily_revenue.get)
print(f"\nBest Selling Day: {best_day} (₹{daily_revenue[best_day]:.2f})")

# 6. Numbered list of all orders using enumerate
print("\n=== All Orders (Numbered List) ===\n")

order_number = 1
for date in sorted(sales_log.keys()):
    for order in sales_log[date]:
        items_str = ", ".join(order["items"])
        print(f"{order_number}. [{date}] Order #{order['order_id']} — ₹{order['total']:.2f} — Items: {items_str}")
        order_number += 1
```
