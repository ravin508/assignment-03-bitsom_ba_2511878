```python
# Task 1: File Read & Write Basics - Production-grade style with error handling

def main():
    filename = "python_notes.txt"

    # PART A: Write to file
    try:
        # Write initial notes
        notes = [
            "Topic 1: Variables store data. Python is dynamically typed.",
            "Topic 2: Lists are ordered and mutable.",
            "Topic 3: Dictionaries store key-value pairs.",
            "Topic 4: Loops automate repetitive tasks.",
            "Topic 5: Exception handling prevents crashes."
        ]

        with open(filename, 'w', encoding="utf-8") as file:
            for note in notes:
                file.write(note + "\n")

        print("File written successfully.")

        # Append two more lines
        with open(filename, 'a', encoding="utf-8") as file:
            file.write("Topic 6: Functions promote code reusability and modularity.\n")
            file.write("Topic 7: File handling is essential for data persistence.\n")

        print("Lines appended successfully.")

    except Exception as e:
        print(f"Error during write operation: {e}")

    # PART B: Read from file
    try:
        with open(filename, 'r', encoding="utf-8") as file:
            lines = file.readlines()

        print("\nFile Contents:")
        for i, line in enumerate(lines, 1):
            print(f"{i}. {line.strip()}")

        print(f"\nTotal number of lines: {len(lines)}")

        # Keyword search
        keyword = input("\nEnter a keyword to search: ").strip()

        if keyword:
            matches = [line.strip() for line in lines if keyword.lower() in line.lower()]

            if matches:
                print(f"\nFound {len(matches)} matching line(s) for '{keyword}':")
                for match in matches:
                    print(f"   {match}")
            else:
                print(f"No lines found containing the keyword '{keyword}'.")
        else:
            print("No keyword entered.")

    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
    except Exception as e:
        print(f"Error during read operation: {e}")


if __name__ == "__main__":
    print("Starting File Operations Program...\n")
    main()
    print("\nProgram completed.")
```

```python
import requests

def main():
    base_url = "https://dummyjson.com/products"

    print("Starting API Integration Program...\n")

    # Step 1: Fetch and Display 20 Products
    try:
        print("Fetching 20 products...")
        response = requests.get(f"{base_url}?limit=20")
        response.raise_for_status()          # Raise error for bad responses
        data = response.json()

        products = data['products']

        print("\n--- All Products (First 20) ---")
        print(f"{'ID':<4} | {'Title':<30} | {'Category':<15} | {'Price':<8} | {'Rating':<6}")
        print("-" * 80)

        for p in products:
            print(f"{p['id']:<4} | {p['title'][:28]:<30} | {p['category']:<15} | "
                  f"${p['price']:<7} | {p['rating']:<6}")

    except Exception as e:
        print(f"Error fetching products: {e}")
        return

    # Step 2: Filter and Sort (Rating >= 4.5, Price descending)
    try:
        filtered = [p for p in products if p['rating'] >= 4.5]
        filtered.sort(key=lambda x: x['price'], reverse=True)

        print("\n--- Filtered & Sorted Products (Rating >= 4.5) ---")
        print(f"{'ID':<4} | {'Title':<30} | {'Category':<15} | {'Price':<8} | {'Rating':<6}")
        print("-" * 80)

        for p in filtered:
            print(f"{p['id']:<4} | {p['title'][:28]:<30} | {p['category']:<15} | "
                  f"${p['price']:<7} | {p['rating']:<6}")

        print(f"\nTotal products after filter: {len(filtered)}")

    except Exception as e:
        print(f"Error during filter/sort: {e}")

    # Step 3: Fetch Laptops Category
    try:
        print("\nFetching laptops category...")
        response = requests.get(f"{base_url}/category/laptops")
        response.raise_for_status()
        laptops = response.json()['products']

        print("\n--- Laptops ---")
        for laptop in laptops:
            print(f"{laptop['title']} - ${laptop['price']}")

    except Exception as e:
        print(f"Error fetching laptops: {e}")

    # Step 4: POST Request (Add new product)
    try:
        print("\nSending POST request to add a new product...")
        new_product = {
            "title": "My Custom Product",
            "price": 999,
            "category": "electronics",
            "description": "A product I created via API"
        }

        response = requests.post(f"{base_url}/add", json=new_product)
        response.raise_for_status()
        result = response.json()

        print("\n--- POST Response ---")
        print(f"ID Assigned     : {result.get('id')}")
        print(f"Title           : {result.get('title')}")
        print(f"Price           : ${result.get('price')}")
        print(f"Category        : {result.get('category')}")
        print(f"Description     : {result.get('description')}")

    except Exception as e:
        print(f"Error during POST request: {e}")


if __name__ == "__main__":
    main()
    print("\nProgram completed.")
```

```python
import requests


# ==================== PART A: Guarded Calculator ====================
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Error: Cannot divide by zero"
    except TypeError:
        return "Error: Invalid input types"


# ==================== PART B: Guarded File Reader ====================
def read_file_safe(filename):
    try:
        with open(filename, 'r', encoding="utf-8") as file:
            return file.read()
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
        return None
    finally:
        print("File operation attempt complete.")


# ==================== PART C: Robust API Call Helper ====================
def safe_get(url, timeout=5):
    try:
        response = requests.get(url, timeout=timeout)
        response.raise_for_status()          # Raises HTTPError for 4xx/5xx
        return response.json()
    except requests.exceptions.ConnectionError:
        print("Connection failed. Please check your internet.")
        return None
    except requests.exceptions.Timeout:
        print("Request timed out. Try again later.")
        return None
    except Exception as e:
        print(f"Error: {e}")
        return None


def safe_post(url, json_data, timeout=5):
    try:
        response = requests.post(url, json=json_data, timeout=timeout)
        response.raise_for_status()
        return response.json()
    except requests.exceptions.ConnectionError:
        print("Connection failed. Please check your internet.")
        return None
    except requests.exceptions.Timeout:
        print("Request timed out. Try again later.")
        return None
    except Exception as e:
        print(f"Error: {e}")
        return None


# ==================== PART D: Interactive Product Lookup ====================
def product_lookup():
    while True:
        user_input = input("\nEnter a product ID to look up (1–100), or 'quit' to exit: ").strip()

        if user_input.lower() == 'quit':
            print("Exiting product lookup.")
            break

        try:
            product_id = int(user_input)
            if not 1 <= product_id <= 100:
                print("Warning: Product ID must be between 1 and 100.")
                continue
        except ValueError:
            print("Warning: Please enter a valid integer.")
            continue

        # Make API call
        url = f"https://dummyjson.com/products/{product_id}"
        data = safe_get(url)

        if data:
            if 'message' in data and data.get('message') == "Product not found":   # In case of 404 handling
                print("Product not found.")
            else:
                print(f"Title : {data.get('title')}")
                print(f"Price : ${data.get('price')}")
        else:
            print("Failed to fetch product.")


# ==================== Main Program ====================
def main():
    print("Starting Task 3 - Exception Handling Program...\n")

    # Part A: Safe Divide Tests
    print("=== Part A: Safe Divide ===")
    print(safe_divide(10, 2))
    print(safe_divide(10, 0))
    print(safe_divide("ten", 2))
    print()

    # Part B: Safe File Reader Tests
    print("=== Part B: Safe File Reader ===")
    content = read_file_safe("python_notes.txt")
    if content:
        print("File read successfully.")

    read_file_safe("ghost_file.txt")
    print()

    # Part D: Interactive Product Lookup
    print("=== Part D: Product Lookup ===")
    product_lookup()


if __name__ == "__main__":
    main()
    print("\nProgram completed.")
```

```python
import requests
from datetime import datetime


# ==================== Logger Function ====================
def log_error(function_name, error_type, message):
    """Log error to file with timestamp"""
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    log_entry = f"[{timestamp}] ERROR in {function_name}: {error_type} — {message}\n"

    try:
        with open("error_log.txt", "a", encoding="utf-8") as file:
            file.write(log_entry)
        print(f"Logged error: {error_type}")
    except Exception as e:
        print(f"Failed to write to log file: {e}")


# ==================== Main Task 4 Program ====================
def main():
    print("Starting Task 4 - Logging to File...\n")

    # Example 1: Trigger ConnectionError (unreachable URL)
    print("Testing unreachable URL (ConnectionError)...")
    try:
        response = requests.get("https://this-host-does-not-exist-xyz.com/api", timeout=5)
    except requests.exceptions.ConnectionError as e:
        log_error("fetch_from_invalid_url", "ConnectionError", str(e))
    except Exception as e:
        log_error("fetch_from_invalid_url", type(e).__name__, str(e))

    # Example 2: Trigger HTTP 404 error (Product ID 999 doesn't exist)
    print("Testing non-existent product (ID 999)...")
    try:
        response = requests.get("https://dummyjson.com/products/999", timeout=5)

        if response.status_code != 200:
            log_error("lookup_product", "HTTPError",
                     f"{response.status_code} Not Found for product ID 999")
        else:
            print("Product found (unexpected).")
    except Exception as e:
        log_error("lookup_product", type(e).__name__, str(e))

    # Final Step: Read and display the log file
    print("\n=== Contents of error_log.txt ===")
    try:
        with open("error_log.txt", "r", encoding="utf-8") as file:
            print(file.read())
    except FileNotFoundError:
        print("No log file found.")
    except Exception as e:
        print(f"Error reading log file: {e}")


if __name__ == "__main__":
    main()
```
